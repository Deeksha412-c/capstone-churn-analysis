```python
import pandas as pd
import numpy as np
from scipy import stats

df = pd.read_csv("C:\\Users\\HP\\OneDrive\\文档\\GitHub\\internship\\week_8\\capstone-churn-analysis\\data\\cleaned_data.csv")

print("Dataset loaded successfully!")
print("Dataset shape:: ",df.shape)
```

```python
stayed = df[df["Churn"] == 0]["Tenure"]
left = df[df["Churn"] == 1]["Tenure"]

print("Stayed customers :: ",len(stayed))
print("Customers who left :: ",len(left))

print("Average tenure of stayed customers :: ", stayed.mean())
print("Average tenure of customers who left :: ", left.mean())
```

```python
t_statistic, p_value = stats.ttest_ind(
    stayed,
    left,
    equal_var = False
)

print("T-Statistic :: ", t_statistic)
print("P-value :: ", p_value)
```

## Hypothesis Test 1: Customer Tenure vs Churn

### Null Hypothesis (H0)
There is no significant difference in average tenure between customers who stayed and customers who left.

### Alternative Hypothesis (H1)
There is a significant difference in average tenure between customers who stayed and customers who left.

### Test Used
Welch's Independent Samples t-test.

### Results
- Average tenure (Stayed): 40.15 months
- Average tenure (Churned): 6.00 months
- T-statistic: 34.5349
- P-value: 8.9823e-130
- Significance level: 0.05

### Decision
Since the p-value is less than 0.05, we reject the null hypothesis.

### Conclusion
There is a statistically significant difference in average tenure between customers who stayed and customers who left.

### Business Recommendation
The company should strengthen its onboarding process and provide additional support to newly acquired customers to reduce early-stage churn.

## 4. Hypothesis Test Results

### Test 1: Tenure vs Churn
- Test used: Welch's Independent Samples t-test
- Null hypothesis (H0): There is no significant difference in average tenure between retained and churned customers.
- Alternative hypothesis (H1): There is a significant difference in average tenure between the two groups.
- P-value: 8.982326e-130
- Decision: Reject H0.
- Interpretation: Churned customers have significantly lower average tenure than retained customers.

### Test 2: Monthly Charges vs Churn
- Test used: Welch's Independent Samples t-test
- Null hypothesis (H0): There is no significant difference in average monthly charges between retained and churned customers.
- Alternative hypothesis (H1): There is a significant difference in average monthly charges between the two groups.
- P-value: 0.0100792
- Decision: Reject H0.
- Interpretation: Churned customers have significantly higher average monthly charges than retained customers.

### Test 3: Contract Type vs Churn
- Test used: Chi-square Test of Independence
- Null hypothesis (H0): Contract type and customer churn are independent.
- Alternative hypothesis (H1): Contract type and customer churn are associated.
- P-value: 9.587354e-07
- Decision: Reject H0.
- Interpretation: Contract type has a statistically significant association with customer churn.

```python
stayed_charges = df[df["Churn"] == 0]["MonthlyCharges"]
left_charges = df[df["Churn"] == 1]["MonthlyCharges"]

print("Average monthly charges (Stayed):: ", stayed_charges.mean())

print("Average monthly charges (Churned):: ", left_charges.mean())
```

```python
t_statistic_charges, p_value_charges = stats.ttest_ind(
    stayed_charges,
    left_charges,
    equal_var=False
)

print("T-statistics :: ", t_statistic_charges)
print("P-value :: ", p_value_charges)
```

## Hypothesis Test 2: Monthly Charges vs Churn

### Null Hypothesis (H0)
There is no significant difference in average monthly charges between customers who stayed and customers who left.

### Alternative Hypothesis (H1)
There is a significant difference in average monthly charges between customers who stayed and customers who left.

### Test Used
Welch's Independent Samples t-test.

### Results
- Average monthly charges (Stayed): 111.72
- Average monthly charges (Churned): 129.77
- T-statistic: -2.6464
- P-value: 0.01008
- Significance level: 0.05

### Decision
Since the p-value is less than 0.05, we reject the null hypothesis.

### Conclusion
There is a statistically significant difference in average monthly charges between customers who stayed and customers who left.

### Business Recommendation
The company should review pricing strategies, plan affordability, and customer-perceived value to identify possible reasons for higher churn among customers with higher monthly charges.

```python
contract_table = pd.crosstab(
    df["Contract"],
    df["Churn"]
)

print(contract_table)
```

```python
chi2, p_value, dof, expected = stats.chi2_contingency(
    contract_table
)

print("Chi-Square statistics :: ", chi2)
print("P-value :: ", p_value)
print("Degrees of Freedom :: ", dof)
```

## Hypothesis Test 3: Contract Type vs Customer Churn

### Null Hypothesis (H0)
Contract type and customer churn are independent.

### Alternative Hypothesis (H1)
Contract type and customer churn are associated.

### Test Used
Chi-square Test of Independence.

### Contingency Table

| Contract Type | Stayed | Left |
|---|---:|---:|
| Month-to-month | 135 | 35 |
| One year | 178 | 8 |
| Two year | 134 | 10 |

### Results
- Chi-square statistic: 27.7153
- P-value: 9.5873e-07
- Degrees of freedom: 2
- Significance level: 0.05

### Decision
Since the p-value is less than 0.05, we reject the null hypothesis.

### Conclusion
There is a statistically significant association between contract type and customer churn.

### Business Recommendation
The company should focus on month-to-month customers through loyalty benefits, personalized support, and suitable long-term contract incentives.

```python
# Recalculate each test and store its p-value separately

t_stat_tenure, p_value_tenure = stats.ttest_ind(
    stayed, left, equal_var=False
)

t_stat_charges, p_value_charges = stats.ttest_ind(
    stayed_charges, left_charges, equal_var=False
)

chi2, p_value_contract, dof, expected = stats.chi2_contingency(
    contract_table
)

# Create summary table
statistical_summary = pd.DataFrame({
    "Analysis": [
        "Tenure vs Churn",
        "Monthly Charges vs Churn",
        "Contract Type vs Churn"
    ],
    "Test Used": [
        "Welch's Independent Samples t-test",
        "Welch's Independent Samples t-test",
        "Chi-square Test of Independence"
    ],
    "P-value": [
        p_value_tenure,
        p_value_charges,
        p_value_contract
    ]
})

statistical_summary["Decision"] = np.where(
    statistical_summary["P-value"] < 0.05,
    "Reject H0",
    "Fail to reject H0"
)

print(statistical_summary.to_string(index=False))
```
